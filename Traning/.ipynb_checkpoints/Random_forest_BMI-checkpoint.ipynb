```python
from Scripts.Random_forest_nested import *
import pickle
import matplotlib.pyplot as plt
from scipy.stats import randint
import seaborn as sns
```

```python
df=pd.read_excel("../Data_preprocessing/All_2010_P.xlsx")
```

```python
df=df.dropna(subset="ND_Diagnosis")
X,y=df.loc[:,["ADHD_PGRS","Sex","Mother_BMI","Diet_PC2"]],df.ND_Diagnosis
```

```python
X.shape
```

```python
param_distributions = {
    'n_estimators':           [300],
    'max_depth':              randint(2, 6),
    'min_samples_leaf':       [10],
    'max_features':           [0.4, 0.5, 0.6, 'sqrt'],
    'max_samples':            [0.5, 0.6, 0.7, 0.8],
    # class_weight removed
}
dataframe =nested_cross_calibrated_rf(
    X, y,
    Number_CV=5,
    number_split=5,
    n_neighbors_imputer=25,
    param_distributions=param_distributions,
    calibration_method="isotonic")

```

```python
ii = dataframe["interaction_importance"]
floor = 0.10 * ii[["main_effect_i", "main_effect_j"]].values.max()
ii[(ii.main_effect_i > floor) & (ii.main_effect_j > floor)] \
  .sort_values("interaction_ratio", ascending=False).head(10)
```

```python
interaction=interaction_matrix(dataframe)

sns.clustermap(interaction)
```
