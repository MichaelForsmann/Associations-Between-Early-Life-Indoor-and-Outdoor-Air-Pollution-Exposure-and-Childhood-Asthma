```python
from sklearn.metrics import precision_recall_curve, average_precision_score

p, r, _ = precision_recall_curve(y.values, Dataframe["oof_probs"])
ap = average_precision_score(y.values, Dataframe["oof_probs"])

plt.plot(r, p, label=f"model (AP = {ap:.3f})")
plt.axhline(0.149, ls="--", color="gray", label=f"prevalence (AP_baseline = 0.149)")
plt.xlabel("Recall"); plt.ylabel("Precision"); plt.legend()
```

```python
# Precision at clinically relevant operating points
from sklearn.metrics import precision_score, recall_score

# Top 20% highest-risk patients: what's the precision?
threshold_top20 = np.quantile(Dataframe["oof_probs"], 0.80)
preds_top20 = (Dataframe["oof_probs"] >= threshold_top20).astype(int)
print(f"Precision at top 20% flagged: {precision_score(y, preds_top20):.3f}")
print(f"Recall at top 20% flagged:    {recall_score(y, preds_top20):.3f}")
```

```python


paths = [f"Models/asthma_12_year_adjusted_repeat{j}_fold{i}.nc" for i in range(3) for j in range(2)]
trees = [az.from_netcdf(p) for p in paths]

# Stack the parameter posteriors across folds
combined_posterior = xr.concat([t["posterior"].to_dataset() for t in trees], dim="chain",
)

asthma = xr.DataTree.from_dict({"posterior": combined_posterior})
list_parameters=["$NO_x$","$NO_2$","Formaldehyd","Acetone","Acetald",'constrution_year',"Black Carbon","$PM_{2.5}$","Street $NO_2$","Street $NO_x$","Street $PM_{2.5}$","BMI","Gestational age","Mother Education",'LRTI',"Polygenetic risk score","Landuse 1","Landuse 2"]
combined_posterior = combined_posterior.assign_coords(weight_dim_0=list_parameters)
combined_posterior["weight_OR"]=np.exp(combined_posterior["weight"])
```

```python
weights = combined_posterior["weight_OR"].stack(sample=("chain", "draw")).values  # (D, S)

# 50% and 95% credible intervals
q025, q25, q50, q75, q975 = np.quantile(weights, [0.025, 0.25, 0.5, 0.75, 0.975], axis=1)

order = np.argsort(q50)
y = np.arange(len(list_parameters))
labels_sorted = [list_parameters[i] for i in order]
colors = ["C3" if q50[i] > 1 else "C0" for i in order]  # red for risk, blue for protective

fig, ax = plt.subplots(figsize=(7, 6))
for k, i in enumerate(order):
    ax.hlines(y[k], q025[i], q975[i], color=colors[k], lw=1)
    ax.hlines(y[k], q25[i], q75[i], color=colors[k], lw=4)
    ax.plot(q50[i], y[k], "o", color="white",
            markeredgecolor=colors[k], markersize=7, zorder=3)

ax.axvline(1.0, color="k", ls="--", lw=1)

ax.set_yticks(y); ax.set_yticklabels(labels_sorted)
ax.set_xlabel("Odds ratio per 1 SD")
ax.grid(axis="x", alpha=0.3)
plt.tight_layout()
```
