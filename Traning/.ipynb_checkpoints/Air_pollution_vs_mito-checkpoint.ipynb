```python
from Scripts.Random_forest_nested import *
import pickle
import matplotlib.pyplot as plt
from scipy.stats import randint
import seaborn as sns
```

```python
df=pd.read_excel("../Data_preprocessing/All_2010_P.xlsx")
```

```python
df=df.dropna(subset="ND_Diagnosis")
X,y=df.loc[:,["ADHD_PGRS","Sex","Mother_BMI","Diet_PC2"]],df.ND_Diagnosis
```

```python
X.shape
```

```python
param_distributions = {
    'n_estimators':           [300],
    'max_depth':              randint(2, 6),
    'min_samples_leaf':       [10],
    'max_features':           [0.4, 0.5, 0.6, 'sqrt'],
    'max_samples':            [0.5, 0.6, 0.7, 0.8],
    # class_weight removed
}
dataframe =nested_cross_calibrated_rf(
    X, y,
    Number_CV=5,
    number_split=5,
    n_neighbors_imputer=25,
    param_distributions=param_distributions,
    calibration_method="isotonic")

```

```python
M, consistent, direction = interaction_matrix(dataframe, consistency=0.8)


k = M.shape[0]
diag = np.eye(k, dtype=bool)
off = M.values[~diag]
vmax = off.max()
thresh = 0.05 * vmax

show = (M.values > thresh) & consistent.values 

fig, ax = plt.subplots(figsize=(15, 10))
sns.heatmap(M, cmap = 'rocket_r', vmin=0, vmax=vmax,
            ax=ax, cbar_kws={'label': r'$2 \overline{|\phi_{i,j}|}$'})

for i in range(k):
    for j in range(k):
        if show[i, j]:
            val = M.iloc[i, j]
            sgn = '' if direction.iloc[i, j] > 0 else '\u2212'
            ax.text(j+0.5, i+0.5, f'{sgn}{val:.4f}', ha='center', va='center',
                    fontsize=8, fontweight='bold',color='white' if val > 0.6*vmax else 'black')
        elif i==j:
            val = M.iloc[i, j]
            sgn = '' if direction.iloc[i, j] > 0 else '\u2212'
            ax.text(j+0.5, i+0.5,f'{sgn}{val:.4f}', ha='center', va='center',
                    fontsize=8, fontweight='bold',
                    color='white')           
ax.set_xticklabels(ax.get_xticklabels(), fontsize=14, fontweight='bold')
ax.set_yticklabels(ax.get_yticklabels(), fontsize=14, fontweight='bold')
cbar = ax.collections[0].colorbar
cbar.ax.yaxis.label.set_size(18)
fig.savefig("../Plots/interactions_BMI.png", dpi=300, bbox_inches='tight')
```

```python
with open("../Performence/results_rf_BMI.pkl", "wb") as f:
    pickle.dump(dataframe, f)
```
